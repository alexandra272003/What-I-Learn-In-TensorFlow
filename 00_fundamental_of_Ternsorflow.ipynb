### Getting information from tensors

When dealing with tensors you probably went to be aware of the following attributes:
* Shape
* Rank
* Axis or Dimensions
* Size

```python
# Create a rank 4 Tensor (4 dimensions)
rank_4_tensor = tf.zeros(shape=[2,3,4,5])
# 2 first dimension , 3 second dimension , 4 `x` element , 5 `y` element

'''
2: This is the number of major groups or 'blocks'.
3: Inside each major group, there are 3 rows or 'sub-groups'.
4: Inside each row, there are 4 columns or 'vectors'.
5: Inside each column (vector), there are 5 individual elements.
'''

rank_4_tensor
```

```python
tf.size(rank_4_tensor), rank_4_tensor.ndim , rank_4_tensor.shape
```

```python
# Get various attributes of our tensor

print("Datatype of every element :", rank_4_tensor.dtype)
print("Number of dimension (rank) :", rank_4_tensor.shape)
print("Shape of Tensor :", rank_4_tensor.ndim)
print("Element along the 0 axis :", rank_4_tensor.shape[0])
print("Element along last axis :", rank_4_tensor.shape[-1])
print("Total number of element in our tensor :", tf.size(rank_4_tensor))
print("Total number of element in our tensor in simple words:", tf.size(rank_4_tensor).numpy())
```

### Indexing and Expanding tensors

### Matrix Multiplication

in machine learning matrix multiplication is one of the most common tensor operation

### There are two rules:

1. inner dimensions must match
2. the resulting matrix has the shape of the outer dimensions

```python
# matrix multiplicaton

print(tensor,"\n")

tf.matmul(tensor,tensor) # matric multiplication
```

```python
tensor * tensor  # element wise
```

```python
mat_1 = tf.constant([[1,2,5],
                      [7,2,1],
                      [3,3,3]])

mat_2 = tf.constant([[3,5],
                      [6,7],
                      [1,8]])

tf.matmul(mat_1,mat_2)
```

```python
# matrix multiplication with python operator "@"

mat_1 @ mat_2
```

```python
# create a tensor (3,2)

x = tf.constant([[1,2],
                  [3,4],
                  [5,6]])

y = tf.constant([[7,8],
                  [9,10],
                  [11,12]])
x,y
```

```python
 # lets change the shape of Y

tf.reshape(y, shape=(2,3))
```

```python
# try to multiply x by reshape y

x @ tf.reshape(y,shape=(2,3))
```

```python
# try tf.matmul

tf.matmul(x , tf.reshape(y,shape=(2,3)))
```

```python
# now change the X

tf.matmul(tf.reshape(x,shape=(2,3)),y)

```

### The difference between `tf.transpose` and `tf.reshape`

*   **`tf.transpose()`**: Permutes the dimensions of a tensor, effectively swapping or reordering its axes. The elements themselves remain the same, but their arrangement changes based on the new dimension order.

*   **`tf.reshape()`**: Changes the shape of a tensor by reinterpreting its elements into a new configuration of dimensions, without changing the order of the elements. The total number of elements must remain constant.

```python
# Can do the same with transpose

x, tf.transpose(x) , tf.reshape(x,shape=(2,3))
```

```python
# try matric multiplication with transopose rather than reshape

tf.matmul(tf.transpose(x),y)
```

## Aggregating Tensors

aggregating tensors = condensing them from multiple values down to a smaller amount of values


```python
# get the absolute values

d = tf.constant([-7,-10])
d
```

```python
# getting the absolute values

tf.abs(d)
```

Lets go through the following forms of aggregation:

* get the minimum
* get the maximum
* get the mean of a tensor
* get the sum of the tensor

```python
# creating a random tensor with values 0 , 100 of size 50

e = tf.constant(np.random.randint(0,100,size=50))
e
```

```python
tf.size(e), e.shape , e.ndim
```

```python
# Find the minimun

tf.reduce_min(e)
```

```python
# find the max

tf.reduce_max(e)
```

```python
# find the mean

tf.reduce_mean(e)
```

```python
# find the sum

tf.reduce_sum(e)
```

```python
# find the variance

tf.math.reduce_variance(tf.cast(e, dtype=tf.float32))
```

```python
# find the st. dev

tf.math.reduce_std(tf.cast(e, dtype=tf.float32))
```

### FIND the positional maximum and minimum

```python
# create a new tensor for finding positional minimum and maximum

tf.random.set_seed(42)
f = tf.random.uniform(shape=[50])
f
```

```python
# find the positional maximum
tf.argmax(f) # this return max index
```

```python
# index on our largest value position

f[tf.argmax(f)]
```

```python
# another keyword

tf.reduce_max(f)
```

```python
# find the positional minimum
tf.argmin(f) # this return min index
```

```python
# index on our smalleest value position

f[tf.argmin(f)]
```

```python
# another keyword

tf.reduce_min(f)
```
